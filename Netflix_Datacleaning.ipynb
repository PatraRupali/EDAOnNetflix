```python
def unstack_col(df, col):
    """
    This function unpacks a comma-separated column (e.g., director, cast, country, listed_in) into individual rows.
    Each value in the column is split and expanded into separate rows, while the 'title' remains as the index.

    Parameters:
    df (DataFrame): The input DataFrame containing the column to unstack.
    col (str): The name of the column to unstack (e.g., 'director', 'cast', etc.).

    Returns:
    DataFrame: A new DataFrame with unstacked values in the specified column.
    """
    Constraint = df[col].apply(lambda x: str(x).split(", ")).tolist()
    df_new1 = pd.DataFrame(Constraint, index=df['title']).stack().reset_index()
    df_new1.drop("level_1", axis=1, inplace=True)
    df_new1.rename({0: col}, axis=1, inplace=True)
    return df_new1
```

```python
#date added column is imputed on the basis of release year,i.e- suppose there's a null for date_added
#when release year was 2013.So below piece of code just checks the mode of date added for release year=2013
# and imputes in place of nulls the corresponding mode
for i in df_final[df_final['date_added'].isnull()]['release_year'].unique():
  imp=df_final[df_final['release_year']==i]['date_added'].mode().values[0]
  
  df_final.loc[df_final['release_year']==i,'date_added']=df_final.loc[df_final['release_year']==i,'date_added'].fillna(imp)
```

```python
df_final.isnull().sum()
```

```python
# Filling missing 'country' values by using the most common country (mode) associated with the same director,
# if the director has other entries with a known country in the dataset.
known_directors = df_final[~df_final['country'].isnull()]['director'].unique()

for director in df_final[df_final['country'].isnull()]['director'].unique():
    if director != 'Unknown Director' and director in known_directors:
        mode_country = df_final[df_final['director'] == director]['country'].mode()
        if not mode_country.empty:
            df_final.loc[(df_final['director'] == director) & (df_final['country'].isnull()), 'country'] = mode_country[0]
```

```python
# Check the number of missing values in each column of the final DataFrame
df_final.isnull().sum()
```

```python
# Imputing missing 'country' values based on the most frequent country associated with the same cast member
known_casts = df_final[~df_final['country'].isnull()]['cast'].unique()

for cast in df_final[df_final['country'].isnull()]['cast'].unique():
    if cast != 'Unknown Actor' and cast in known_casts:
        mode_cast = df_final[df_final['cast'] == cast]['country'].mode()
        if not mode_cast.empty:
            df_final.loc[(df_final['cast'] == cast) & (df_final['country'].isnull()), 'country'] = mode_cast[0]

```

```python
# Titles like 'Boomika (Telugu)', 'Boomika (Malayalam)', 'Boomika (Hindi)' are treated as the same movie in different languages.
# The content inside brackets (e.g., the language) is removed to ensure consistency in the title.
# This operation also trims any leading or trailing spaces from the title.
df_final['title'] = df_final['title'].str.replace(r"\s*\(.*\)\s*", "", regex=True).str.strip()
```

```python
# This code processes the 'date_added' column in the DataFrame by:

from dateutil.parser import parse
# Step 1:Converting it to a string and stripping any extra spaces.
df_final['date_added'] = df_final['date_added'].astype(str).str.strip()  # Convert to string and strip

# Step 2: Convert to datetime safely (in case of any invalid date format)
df_final['date_added'] = pd.to_datetime(df_final['date_added'], errors='coerce')

# Step 3: Extracting the month, week, and year from the 'date_added' column
month_added = df_final['date_added'].dt.month
week_added = df_final['date_added'].dt.isocalendar().week
year_added= df_final['date_added'].dt.year

#Step 4. Inserting the extracted values as new columns: 'month_added', 'week_added', and 'year_added'.
df_final.insert(7,'month_added',month_added)
df_final.insert(8,'week_added',week_added)
df_final.insert(9,'year_added',year_added)
```
